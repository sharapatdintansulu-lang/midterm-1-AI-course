```python
import math, pandas as pd
ranks = pd.read_csv("my_data/retrieval_ranks.csv")

def wilson(h, n, z=1.96):
    p = h/n; d = 1 + z**2/n
    c = (p + z**2/(2*n))/d
    w = z*math.sqrt(p*(1-p)/n + z**2/(4*n**2))/d
    return c-w, c+w

print("n =", len(ranks), ranks.lang.value_counts().to_dict())

# 1a
a = []
for cfg in "AB":
    r = ranks[f"rank_{cfg}"]
    a.append([cfg, (r<=1).mean(), (r<=3).mean(), (r<=5).mean(),
              sum(1/x for x in r if x<=10)/len(r)])
print("\n1a"); print(pd.DataFrame(a, columns=["config","r@1","r@3","r@5","MRR@10"]).round(4).to_string(index=False))


c = []
for cfg in "AB":
    r = ranks[f"rank_{cfg}"]
    for k in (3,5):
        h = int((r<=k).sum()); lo, hi = wilson(h, len(r))
        c.append([f"{cfg},k={k}", h, round(h/len(r),4), round(lo,4), round(hi,4)])
c = pd.DataFrame(c, columns=["cand","hits","recall","lower","upper"])
print("\n1b"); print(c.to_string(index=False))
print("\nintersection of intervals:")
for i in range(4):
    for j in range(i+1,4):
        x, y = c.iloc[i], c.iloc[j]
        ov = x.lower <= y.upper and y.lower <= x.upper
        print(x.cand, "vs", y.cand, "->", "intersect" if ov else "do not intersect")


rows = []
for cfg in "AB":
    for k in (3,5):
        for l, g in ranks.groupby("lang"):
            h = int((g[f"rank_{cfg}"]<=k).sum()); lo, hi = wilson(h, len(g))
            rows.append([f"{cfg},k={k}", l, h, len(g), round(h/len(g),3), round(lo,3), round(hi,3)])
print("\nby language"); print(pd.DataFrame(rows, columns=["cand","lang","hits","n","recall","lower","upper"]).to_string(index=False))
```
